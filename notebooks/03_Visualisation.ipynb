# Visualizations

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("itunes_EDA_cleaned_analytics_3.csv")
```

### First Visualization: Track Duration Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['track_time_millis'] / 60000, bins=40)
plt.xlabel("Track Duration (Minutes)")
plt.ylabel("Frequency")
plt.title("Distribution of Track Duration")
plt.show()
```

#### Chart 2 — Price Distribution

```python

plt.figure(figsize=(7,5))
df['track_price'].value_counts().sort_index().plot(kind='bar')

plt.xlabel("Track Price ($)")
plt.ylabel("Number of Tracks")
plt.title("Distribution of Track Prices")
plt.xticks(rotation=0)
plt.show()
```

### Chart 3 — Duration vs Price

```python

plt.figure(figsize=(8,5))
df.boxplot(column='track_time_millis', by='track_price')

plt.xlabel("Track Price ($)")
plt.ylabel("Track Duration (MilliSecs)")
plt.title("Track Duration by Price Tier")
plt.suptitle("")  # removes automatic extra title
plt.show()
```

### Chart 4 — Top 10 Genres by Volume

```python

top_genres = df['genre'].value_counts().head(10)

plt.figure(figsize=(9,5))
top_genres.plot(kind='bar')

plt.xlabel("Genre")
plt.ylabel("Number of Tracks")
plt.title("Top 10 Genres by Track Volume")
plt.xticks(rotation=45)
plt.show()
```

Bollywood and Pop have the highest number of tracks.

There is a big drop in track count after the top two genres.

Most tracks belong to a few popular genres.

Smaller genres have much fewer tracks.

The catalog is focused more on mainstream genres than niche ones.

### Chart 5 — Track Duration by Genre

```python

top_10_genres = df['genre'].value_counts().head(10).index

plt.figure(figsize=(10,6))
df[df['genre'].isin(top_10_genres)].boxplot(
    column='track_time_millis',
    by='genre',
    rot=45
)

plt.xlabel("Genre")
plt.ylabel("Track Duration (MilliSecs)")
plt.title("Track Duration by Genre")
plt.suptitle("")
plt.show()
```

Most genres have a similar average duration (around 3–4 minutes).

Bollywood and Soundtrack show slightly higher median durations compared to other genres.

Punjabi Pop and Country appear to have slightly shorter median durations.

Pop and Bollywood show many high-duration outliers (very long tracks).

Overall, track length does not vary drastically by genre — durations are mostly standardized across genres.
