## Detect Duration Outliers 

```python
df[df['track_time_millis'] > 600000][['track_name', 'track_time_millis']].sort_values(by='track_time_millis', ascending=False).head(40)
```

#### Insight-1  — Dataset Is Not Pure “Songs”

This dataset contains:

-Non-stop remix compilations

-Workout mixes

-Devotional chants

-Long-form audio


#### Insight 2 — Why This Matters

If we calculate:

    -Average duration by genre

    -Median duration

These long tracks will distort mean heavily.

So for fair music analysis, we may need:

    A filtered dataset (tracks < 10 minutes)

### How Many Tracks Are > 10 Minutes?


```python
(df['track_time_millis'] > 600000).sum()
```

Only 14 tracks are longer than 10 minutes.

Out of ~4,913 tracks, that is:
≈0.28%

So these are true outliers, not structural bias.

### Create Clean Dataset having less than 10 min duration song tracks

```python
df_clean = df[df['track_time_millis'] <= 600000]
df_clean['track_time_millis'].describe()
```

1] Mean dropped
    From 3.99 min → 3.92 min
    
2] Standard deviation cut almost in half
    120k → 66k

    That means:
    ✔ Distribution is now tighter
    ✔ Less skew

3] Maximum now realistic

    588,411 ms ≈ 9.8 minutes
    This is acceptable for long songs.

## checking: Album Price Consistency & Album Size Distribution

Does the same album (collection_name) appear with multiple different collection_price values?

```python
album_validation = (
    df.groupby('collection_name')
      .agg(
          tracks_per_album=('track_id', 'count'),
          unique_collection_prices=('collection_price', 'nunique'),
          collection_price=('collection_price', 'first')
      )
      .reset_index()
)

album_validation.head()
```

```python
print("Albums with multiple collection prices:",
      (album_validation['unique_collection_prices'] > 1).sum())

print("\nMean tracks per album:",
      album_validation['tracks_per_album'].mean())

print("Albums with 1 track:",
      (album_validation['tracks_per_album'] == 1).sum())

print("Albums with 100+ tracks:",
      (album_validation['tracks_per_album'] >= 100).sum())

max_price = album_validation['collection_price'].max()

print("\nMax Collection Price:", max_price)

print("Track count for max price album:")
print(
    album_validation[
        album_validation['collection_price'] == max_price
    ][['tracks_per_album', 'collection_price']]
)
```
