#### MISSING VALUE check

```python
df.isna().sum()
```

```python
df[['track_price','collection_price','track_time_millis','release_date']].describe()
```

```python
# Replace -1 with NaN
df['track_price'] = df['track_price'].replace(-1, pd.NA)
df['collection_price'] = df['collection_price'].replace(-1, pd.NA)

# Check again
df[['track_price','collection_price']].describe()
df[['track_price','collection_price']].isnull().sum()
```

```python
df[['track_price','collection_price']].describe()
```

```python
df[['track_price','collection_price','track_time_millis','release_date']].isnull().sum()
```

Checking If duplicated rows present 

```python
df.duplicated()
```

```python
df.info()
```

there are 4 columns having missing values.

* **artist_name (2 missing)** → Very small number. It won’t affect the analysis.

* **track_price & collection_price (~8% missing)** → We don’t want to guess or fill prices because that would create fake data. We can analyze prices using the available values only.

* **release_date (1% missing)** → Very few missing, and we can ignore them when doing time-based analysis.

In short:
The missing values are either very small in number or should not be filled because it would create incorrect data.
